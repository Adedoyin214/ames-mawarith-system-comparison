# Mawārith Analytics — Project 5
## Cross-System Comparative Analysis of Inheritance Distribution
**Systems:** Islamic (Fiqh al-Mawārith) · Nigerian Statutory (AEL) · Yoruba Customary · Igbo Customary · Civil Equal-Share  
**Dataset:** 20 PCIED cases — Alhikma University, Ilorin  
**Institute:** Elbaseety Institute® | Mawārith Analytics


```python
import json, statistics as st
from comparison_engine import compare_all_cases, SYSTEMS, gini

comparisons = compare_all_cases()
data = json.load(open('comparison_results.json'))
print(f"Loaded {len(comparisons)} cases across {len(SYSTEMS)} systems")
print("\nSystems compared:")
for s in SYSTEMS: print(f"  • {s}")
```

## 1. Aggregate Metrics by System

```python
print(f"{'System':<26} {'Mean Gini':>10} {'Mean Coverage':>14} {'Mean Assigned':>14}")
print("-"*68)
for s in SYSTEMS:
    ginis = [c['systems'][s]['gini'] for c in data]
    cov   = [c['systems'][s]['coverage_rate'] for c in data]
    assn  = [c['systems'][s]['total_assigned']/c['total_value']*100 for c in data]
    print(f"{s:<26} {st.mean(ginis):>10.4f} {st.mean(cov):>13.1f}% {st.mean(assn):>13.1f}%")
```

## 2. Key System Profiles

```python
profiles = {
    "Islamic": "Structured differential shares (1/2, 1/4, 1/6, 1/8) based on relationship & gender. 'Awl/Radd ensure sum=1. Mean coverage 69.3% — blocked heirs excluded.",
    "Nigerian Statutory": "Spouse gets 1/3 (children present) or 1/2. Children share equally regardless of gender. Mean coverage 37.1% — only primary heirs included.",
    "Yoruba Customary": "Eldest son (Dawodu) gets double share. Daughters get half of each son's share. Wife gets usufruct only (~10%). Mean coverage 38.2%.",
    "Igbo Customary": "Sons inherit fully; first son (Okpara) gets 1.5x. Daughters traditionally receive nothing. Wife gets maintenance only (~5%). Mean coverage 20.6% — most exclusive system.",
    "Civil Equal-Share": "Every heir gets exactly 1/N of estate. Gini=0 by construction. Coverage always 100% — the most inclusive benchmark.",
}
for s, desc in profiles.items():
    print(f"\n{s}:\n  {desc}")
```

## 4. Daughter Share Comparison

```python
print("DAUGHTER SHARE — % of estate received by daughter group, per case per system\n")
print(f"{'Case':>4}  {'Islamic':>10}  {'Statutory':>10}  {'Yoruba':>10}  {'Igbo':>8}  {'Civil':>8}")
print("-"*56)
for c in data:
    row = [c['case_id']]
    for s in SYSTEMS:
        d_shares = [v for k,v in c['systems'][s]['shares'].items() if k.startswith('daughter') and v>0]
        total_d = sum(d_shares) / c['total_value'] * 100 if d_shares else 0
        row.append(total_d)
    print(f"  C{row[0]:<2}  {row[1]:>9.1f}%  {row[2]:>9.1f}%  {row[3]:>9.1f}%  {row[4]:>7.1f}%  {row[5]:>7.1f}%")
```

## 5. Key Findings

| Dimension | Winner | Worst |
|-----------|--------|-------|
| **Lowest inequality (Gini)** | Civil Equal-Share (0.0000) | Islamic (0.2295) |
| **Highest coverage** | Civil Equal-Share (100%) | Igbo Customary (20.6%) |
| **Estate fully distributed** | Islamic & Civil (100%) | Igbo/Yoruba (partial) |
| **Best for daughters** | Civil / Islamic | Igbo (0% in many cases) |
| **Best for wives** | Islamic | Igbo (maintenance only) |
| **Most nuanced** | Islamic | Igbo (binary male/female) |
| **Cases 13 & 14 only** | Islamic & Civil | Statutory/Yoruba/Igbo = ₦0 |

> **Note:** Islamic produces the *highest* Gini because it is the most nuanced system — differentiating between the most heir types. A higher Gini is not necessarily worse; it reflects intentional structural differentiation.
